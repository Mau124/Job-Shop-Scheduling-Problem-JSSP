# Test Solvers

This notebook has the intention to test the solvers with a basic problem instance

```python
from pathlib import Path
import sys

# Automatically finds your project root (Job-Shop-Scheduling) from the notebooks folder
project_root = Path().resolve().parent
src_path = project_root / "src"

# Force Python to recognize the src folder
if str(src_path) not in sys.path:
  sys.path.insert(0, str(src_path))

print("Path added successfully:", src_path)
```

```python
from jssp.solvers.recursive import DFSSolver
from jssp.solvers.astar import AStarSolver
from jssp.solvers.csp import CSPSolver
from jssp.core.utils import plot_jssp_schedule
from jssp.core.utils import read_problem_instance
from jssp.solvers.tabu import TabuSolver
```

```python
path_problem = '../instances/benchmarks/toy/ft06.json'
instance = read_problem_instance(path_problem)
seed = 42

path_easy_problem = '../instances/generated/simple/example3.json'
easy_instance = read_problem_instance(path_easy_problem)
```

## DFS Solver

```python
solver = DFSSolver(easy_instance, {})
solver.solve()
result = solver.result
print(result)
```

```python
plot_jssp_schedule(result.schedule)
```

## A* Solver

### Heuristic h1

```python
solver = AStarSolver(easy_instance, hyperparameters={'heuristic': 'h1'})
solver.solve()
result = solver.result
print(result)
```

```python
plot_jssp_schedule(result.schedule)
```

### Heuristic h2

```python
solver = AStarSolver(easy_instance, hyperparameters={'heuristic': 'h2'})
solver.solve()
result = solver.result
print(result)
```

```python
plot_jssp_schedule(result.schedule)
```

## Tabu Solver

### Neighbourhood n1

```python
solver = TabuSolver(instance, seed, hyperparameters={'neighbourhood': 'n1'})
solver.solve()
result = solver.result
print(result)
```

```python
plot_jssp_schedule(result.schedule)
```

### Neighbourhood n2

```python
solver = TabuSolver(instance, seed, hyperparameters={'neighbourhood': 'n2'})
solver.solve()
result = solver.result
print(result)
```

```python
plot_jssp_schedule(result.schedule)
```

## CSP Solver

```python
solver = CSPSolver(easy_instance, hyperparameters={})
solver.solve()
result = solver.result
print(result)
```

```python
plot_jssp_schedule(result.schedule)
```
